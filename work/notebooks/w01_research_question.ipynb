## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content items, and all 30,000 items have at least one impression in the 90-day window.
```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Number of content items:", len(df))

print("\nCTR statistics:")
print(df["ctr"].describe())

print("\nEngagement rate statistics:")
print(df["engagement_rate"].describe())

print("\nImpressions (90d) statistics:")
print(df["impressions_90d"].describe())

print("\nContent with at least 1 impression:")
print((df["impressions_90d"] > 0).sum())

print("\nPercentage with at least 1 impression:")
print((df["impressions_90d"] > 0).mean() * 100)
```
